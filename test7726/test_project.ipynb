```python
import tensorflow as tf

# Data shape rule: (Samples, Features)
dense_model = tf.keras.Sequential(
    [
        tf.keras.layers.Input(shape=(10,)),  # 10 independent features
        tf.keras.layers.Dense(32, activation="relu"),
        tf.keras.layers.Dense(1),
    ]
)
print("Dense Model Built Successfully!")
```

```python
# Data shape rule: (Samples, Time Steps, Features)
rnn_model = tf.keras.Sequential(
    [
        tf.keras.layers.Input(
            shape=(5, 1)
        ),  # Look back 5 time steps, 1 feature per step
        tf.keras.layers.LSTM(
            32, return_sequences=False
        ),  # Swap to .GRU() instantly if needed
        tf.keras.layers.Dense(1),
    ]
)
print("LSTM Model Built Successfully!")
```

```python
# Functional API setup
inputs = tf.keras.layers.Input(shape=(20, 1))  # 20 time steps
lstm_seq = tf.keras.layers.LSTM(32, return_sequences=True)(inputs)  # Must be True

# Attention mechanism weighs the steps
attention_out = tf.keras.layers.Attention()([lstm_seq, lstm_seq])

pooled = tf.keras.layers.GlobalAveragePooling1D()(attention_out)
outputs = tf.keras.layers.Dense(1)(pooled)

attention_model = tf.keras.Model(inputs=inputs, outputs=outputs)
print("Attention Model Built Successfully!")
```

```python
# Minimal Transformer Block architecture
inputs = tf.keras.layers.Input(shape=(15, 64))  # (Sequence Length, Feature Dimensions)

# 1. Multi-Head Self-Attention Layer
attn_output = tf.keras.layers.MultiHeadAttention(num_heads=4, key_dim=64)(
    inputs, inputs
)
out1 = tf.keras.layers.LayerNormalization()(inputs + attn_output)  # Residual connection

# 2. Feed-Forward Network Layer
ffn_output = tf.keras.layers.Dense(64, activation="relu")(out1)
final_out = tf.keras.layers.LayerNormalization()(out1 + ffn_output)

transformer_block = tf.keras.Model(inputs=inputs, outputs=final_out)
print("Transformer Block Built Successfully!")
```
